### Importing Libraries

### Pulling API_KEY from .env & Validating

```python
load_dotenv()
api_key = os.getenv("API_KEY")

print(f'API_KEY present:{bool(api_key)}')
```

### Small Array Op

```python
ar = np.array([x for x in range(1,26) if x%2==1])
dic = {
    'mean':int(ar.mean()),
    'max':int(ar.max()),
    'min':int(ar.min()),
    }
print(dic)
```
